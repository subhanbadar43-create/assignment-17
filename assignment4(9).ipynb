```python
def validtransaction(balance, amount, dailylimit, usedtoday):
    if amount <= 0:
        print("Transaction Rejected")
        print("Reason: Amount must be greater than 0")
    elif amount > balance:
        print("Transaction Rejected")
        print("Reason: Amount exceeds account balance")
    elif (usedtoday + amount) > dailylimit:
        print("Transaction Rejected")
        print("Reason: Daily transaction limit exceeded")
    elif (balance - amount) < 1000:
        print("Transaction Rejected")
        print("Reason: Remaining balance would be below Rs. 1000")
    else:
        remainingbalance = balance - amount
        remaininglimit = dailylimit - usedtoday - amount
        print("Transaction Approved")
        print("Remaining Balance:", remainingbalance)
        print("Remaining Daily Limit:", remaininglimit)
    print("------------------------")
validtransaction(50000, 10000, 20000, 5000)
validtransaction(50000, 0, 20000, 5000)
validtransaction(5000, 10000, 20000, 5000)
validtransaction(50000, 16000, 20000, 5000)
validtransaction(1500, 800, 20000, 0)
```
